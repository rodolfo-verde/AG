```python
#Sucesión_de_Fibonacci
#https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci
#Calculo del termino n-simo de la suscesión de Fibonacci
def Fibonacci(N:int):
  if N < 2:
    return 1
  else:
    return Fibonacci(N-1)+Fibonacci(N-2)

Fibonacci(5)
```

```python
import time

#Con memoizacion: guardo en un diccionario lo que ya he calculado para
#  no recalcular los mismos terminos
def Fibonacci_memo(N, memoria):
  if N in memoria:
    return memoria[N]
  if N < 2:
    memoria[N] = 1
    return 1
  memoria[N] = Fibonacci_memo(N-1, memoria) + Fibonacci_memo(N-2, memoria)
  return memoria[N]

t0 = time.time()
print("Fibonacci(30) recursivo :", Fibonacci(30), " en %.4f s" % (time.time()-t0))

t0 = time.time()
memoria = {}
valor = Fibonacci_memo(950, memoria)
print("Fibonacci_memo(950)     :", len(str(valor)), "digitos en %.6f s" % (time.time()-t0))

#  (con 950 me pillo el limite de recursividad de python, con la memoizacion
#   podria ir mucho mas lejos)
```

##Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

```python
#El cuaderno imprime todas las soluciones; aqui las cuento
#  (reutilizo es_prometedora del cuaderno)
def contar_reinas(N, solucion=None, etapa=0):
  if solucion is None:
    solucion = [0]*N
  if etapa == N:
    return 1
  total = 0
  for col in range(1, N+1):
    solucion[etapa] = col
    if es_prometedora(solucion, etapa):
      total += contar_reinas(N, solucion, etapa+1)
    solucion[etapa] = 0
  return total

print("N=4:", contar_reinas(4), "soluciones")
print("N=8:", contar_reinas(8), "soluciones")
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

```python
#Al calcular_ruta del cuaderno le falta el destino al final (y le sobra
#una coma). Lo dejo devolviendo la lista completa y compruebo que la suma
#de los tramos da el precio optimo
def ruta_completa(RUTA, desde, hasta):
  ruta = [hasta]
  actual = hasta
  while RUTA[desde][actual] != desde:
    actual = RUTA[desde][actual]
    ruta.append(actual)
  return [desde] + ruta[::-1]

ruta = ruta_completa(RUTA, 0, 6)
print("La ruta completa es:", ruta)
print("Suma de tramos:", sum(TARIFAS[a][b] for a,b in zip(ruta, ruta[1:])),
      " (precio optimo:", PRECIOS[0][6], ")")
```

##Actividad extra: par de puntos mas cercano
En 1D lo hago por fuerza bruta (O(n^2)) y por dividir y venceras
(O(n log n)), comprobando que ambos dan el minimo de verdad. Despues extiendo
el algoritmo a 2D y 3D: se hace un corte en el plano y los pares que lo
cruzan se buscan en la franja de anchura 2*delta.


```python
import random
import time
import math

#ACTIVIDAD EXTRA (los 2 puntos de la nota hasta el 10): el par de puntos
#mas cercano. La clase: empezar en 1D por fuerza bruta (complejidad muy
#alta), mejorarlo con dividir y venceras, y extenderlo a 2D y 3D.

#--- 1D ---------------------------------------------------------------

#Fuerza bruta: pruebo todos los pares -> O(n^2)
def par_cercano_1D_fb(v):
  mejor_d = float("inf")
  par = None
  for i in range(len(v)):
    for j in range(i+1, len(v)):
      d = abs(v[i]-v[j])
      if d < mejor_d:
        mejor_d = d
        par = (v[i], v[j])
  return par, mejor_d


#Divide y venceras: ordeno una vez (O(n log n)) y busco recursivamente el
#  par mas cercano en la mitad izquierda y en la derecha. En 1D ordenado
#  el unico par que cruza la mediana son los dos elementos de enmedio.
#  T(n) = 2*T(n/2) + O(1)  ->  O(n log n)
def par_cercano_1D_dyv(v):
  v = sorted(v)

  def dyv(i, j):  #trabaja sobre v[i..j]
    if j - i == 0:  #un punto: no hay par
      return None, None, float("inf")
    if j - i == 1:  #dos puntos: el par ya esta
      return v[i], v[j], v[j]-v[i]
    m = (i+j)//2
    a, b, d_izq = dyv(i, m)
    c, d, d_der = dyv(m+1, j)
    if d_izq <= d_der:
      a, b, d = a, b, d_izq
    else:
      a, b, d = c, d, d_der
    #El par que cruza la mediana
    if v[m+1]-v[m] < d:
      a, b, d = v[m], v[m+1], v[m+1]-v[m]
    return a, b, d

  a, b, d = dyv(0, len(v)-1)
  return (a, b), d


#Comprobar que el dividir y venceras da el minimo de verdad (y no una
#  solucion suboptima): lo contrasto con la fuerza bruta
random.seed(42)
for n in [10, 50, 200]:
  v = [random.randint(0, 1000) for _ in range(n)]
  par_fb, d_fb = par_cercano_1D_fb(v)
  par_dyv, d_dyv = par_cercano_1D_dyv(v)
  print(n, "| fuerza bruta:", d_fb, "| dividir y venceras:", d_dyv, "| iguales:", d_fb == d_dyv)

#La mejora de complejidad se nota en el tiempo
v = [random.randint(0, 1000000) for _ in range(2000)]
t0 = time.time(); par_fb, d_fb = par_cercano_1D_fb(v); t_fb = time.time() - t0
t0 = time.time(); par_dyv, d_dyv = par_cercano_1D_dyv(v); t_dyv = time.time() - t0
print()
print("n=2000 -> fuerza bruta %.3f s | dividir y venceras %.3f s" % (t_fb, t_dyv))
print("minimo distancia:", d_fb)
```

```python
#Extender el algoritmo a 2D y 3D: mismo concepto (corte + dividir y
#  venceras), pero ahora los pares que cruzan el corte se buscan en una
#  franja central de anchura 2*delta.

def dist2(p, q):
  return sum((a-b)**2 for a, b in zip(p, q))


def par_cercano_ND(puntos):
  #Ordeno por la primera coordenada y hago el corte por la mitad.
  #La recursion da el mejor par de cada mitad (delta = sqrt(d2)).
  #Solo faltan los pares que cruzan el corte: miro la franja central
  #  (|x - xm| <= delta) ordenada por la segunda coordenada y solo
  #  compruebo los pares con |dy| <= delta. En 2D los candidatos por
  #  punto estan acotados (7); en 3D no hay tan buen bound, pero con la
  #  franja y la ventana en y las comprobaciones quedan muy por debajo
  #  de O(n^2).
  P = sorted(puntos, key=lambda p: p[0])

  def dyv(P):
    n = len(P)
    if n <= 3:  #tan pocos puntos que la fuerza bruta es mas rapida
      mejor = float("inf"); par = None
      for i in range(n):
        for j in range(i+1, n):
          d = dist2(P[i], P[j])
          if d < mejor:
            mejor = d; par = (P[i], P[j])
      return par, mejor
    m = n//2
    (a, b), d_izq = dyv(P[:m])
    (c, d), d_der = dyv(P[m:])
    if d_izq <= d_der:
      par, d2 = (a, b), d_izq
    else:
      par, d2 = (c, d), d_der
    #Franja central: |x - xm| <= delta
    xm = P[m-1][0]
    franja = [p for p in P if (p[0]-xm)**2 <= d2]
    franja.sort(key=lambda p: p[1])
    for i in range(len(franja)):
      j = i + 1
      while j < len(franja) and (franja[j][1]-franja[i][1])**2 <= d2:
        d = dist2(franja[i], franja[j])
        if d < d2:
          d2 = d; par = (franja[i], franja[j])
        j += 1
    return par, d2

  par, d2 = dyv(P)
  return par, math.sqrt(d2)


#Comprobar contra la fuerza bruta con nubes de puntos aleatorias
random.seed(1)
for dim in (2, 3):
  for n in [10, 50, 200]:
    pts = [tuple(random.randint(0, 500) for _ in range(dim)) for _ in range(n)]
    mejor = float("inf")
    for i in range(n):
      for j in range(i+1, n):
        d = dist2(pts[i], pts[j])
        if d < mejor:
          mejor = d
    par2, d2 = par_cercano_ND(pts)
    print(dim, "D, n=", n, "| fuerza bruta:", round(math.sqrt(mejor), 3),
          "| dividir y venceras:", round(d2, 3),
          "| iguales:", math.isclose(math.sqrt(mejor), d2))
```
